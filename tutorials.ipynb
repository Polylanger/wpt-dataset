## 1. Loading the Dataset

```python
from dgl import load_graphs

graph_path = 'data/host-1.bin'
graphs, _ = load_graphs(graph_path)
```

## 2. Accessing Graph Data

```python
# Access the first graph snapshot
print("The first graph snapshot: ", graphs[0])

# Check the total number of graph snapshots
num_graphs = len(graphs)
print(f"Total number of graph snapshots: {num_graphs}")
```

## 3. Viewing Graph Properties and Data

In a discrete dynamic graph, if a node appears later on, it is customary to use 0 as a placeholder before its actual appearance. This approach ensures the continuity and integrity of the graph, indicating the absence of the node at that particular moment. Such a representation clearly showcases the emergence sequence and dynamic changes of nodes, facilitating the analysis and understanding of the graph's dynamic characteristics.

```python
g = graphs[0]  # Example using the first snapshot

# Print basic graph info
print(f"Number of nodes: {g.number_of_nodes()}")
print(f"Number of edges: {g.number_of_edges()}")

# Access and print part of the node data
node_feat = g.ndata['feat']  # Node features
node_info = g.ndata['info']  # Additional info
node_label = g.ndata['label']  # Node labels

# Print the shape and dtype of node features and info
print(f"Node features shape: {node_feat.shape}")
print(f"Node info shape: {node_info.shape}")

# Explain the divisions within the node features
print("Node feature vector breakdown:")
print("  Structural features (first 72 dimensions):", node_feat[:, :72])
print("  Textual features (middle 64 dimensions):", node_feat[:, 72:136])
print("  Statistical features (last 12 dimensions):", node_feat[:, 136:])

# Explain the info attribute
print("Node info attribute details:")
info_ids = node_info[:, 0]  # First dimension: Original webpage IDs
info_tampering = node_info[:, 1]  # Second dimension: Tampering types
print(f"  Original IDs (first dimension of info): {info_ids}")
print(f"  Tampering types (second dimension of info): {info_tampering}")

# Mapping of tampering types
tampering_types = {0: "Normal Pages", 1: "Website Homepage Replaced", 2: "Website Homepage Infected with Trojan", 3: "Malicious Ads Inserted in Website"}
mapped_tampering_types = [tampering_types.get(int(t), "Unknown") for t in info_tampering]
print("  Mapped tampering types:", mapped_tampering_types)
```
